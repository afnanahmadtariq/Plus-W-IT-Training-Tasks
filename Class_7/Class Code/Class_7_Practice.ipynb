```python
#linear Regression
import seaborn as sns
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression

#Load the dataset
df = sns.load_dataset('tips')
print(df)
X = df['total_bill'].values.reshape(-1, 1)
print(X)
y = df['tip'].values
print(y)

#Build the relationship between X and Y
model = LinearRegression()
model.fit(X, y)

#Predict the values
residuals = model.predict(X) - y
sm.qqplot(residuals, line='s')

```

```python
import numpy as np
import matplotlib.pyplot as plt

#Sample data
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

#Fit Model
model = LinearRegression()
model.fit(X.reshape(-1, 1), y)

#Plot
plt.scatter(X, y, color='red', label='Data Points')
plt.plot(X, model.predict(X.reshape(-1, 1)), color='blue', label='Regression Line')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()



```

```python
#Build the Linear Regression Model to predict houses prices
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

#Define Data set
square_feet = np.array([1000, 1500, 2000, 2500, 3000])
house_prices = np.array([200000, 250000, 300000, 350000, 400000])

#Build the model / Train the model
model = LinearRegression() #Iintialize the model
model.fit(square_feet.reshape(-1, 1), house_prices) #Train the model

#Make a prediction
predicted_price = model.predict(np.array([2000]).reshape(-1, 1))
print(f'Predicted House price for 2000 sqft is: ${predicted_price}')
```

```python
plt.scatter(square_feet, house_prices, color='red', label='Actual Data')
plt.plot(square_feet, model.predict(square_feet.reshape(-1, 1)), color='blue', label='Regression Line')
plt.xlabel('Square Feet')
plt.ylabel('House Prices')
plt.title('House Price Prediction')
plt.legend()
plt.show()
```

```python
#Task 1
import numpy as np
import matplotlib.pyplot as plt

def linear_function(x, m=2, c=3):
    return m * x + c

#Task 2
x_values = np.linspace(-10, 10, 100)
y_values = linear_function(x_values)

#Task 3
plt.plot(x_values, y_values, label='y = 2x + 3', color='b', linestyle='-', linewidth=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def mse(y_actual, y_predicted):
    return np.mean((y_actual - y_predicted) ** 2)

y_actual = np.array([1, 2, 3, 4, 5])
y_predicted = np.array([2, 3, 4, 5, 6])

mse_value = mse(y_actual, y_predicted)
print(f'Mean Squared Error: {mse_value}')
```
